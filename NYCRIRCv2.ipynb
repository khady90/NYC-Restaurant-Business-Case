## 2019 dataset — initial modelling preparation

```python
#URL = "https://data.cityofnewyork.us/resource/43nn-pn8j.csv?$limit=100000"
restaurant_df = pd.read_csv("2019_New_York_City_Restaurant_Inspection_Results_20260810.csv")
restaurant_df.head()
```

```python
restaurant_df.columns = (
    restaurant_df.columns
    .str.lower()
    .str.replace(" ", "_")
)

restaurant_df["inspection_date"] = pd.to_datetime(
    restaurant_df["inspection_date"],
    errors="coerce"
)

restaurant_df["month"] = restaurant_df["inspection_date"].dt.month
restaurant_df.head()
```
